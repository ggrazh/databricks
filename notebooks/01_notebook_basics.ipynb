```python
data = [
    (1, "Alice", 100),
    (2, "Bob", 200),
    (3, "Charlie", 150)
]

df = spark.createDataFrame(
    data,
    ["id", "customer", "amount"]
)

display(df)
```

```python
from pyspark.sql.functions import col

filtered_df = df.filter(col("amount") > 100)

display(filtered_df)
```

```python
df.createOrReplaceTempView("sales")
```

```python
%sql
SELECT customer, amount
FROM sales
WHERE amount > 100
ORDER BY amount DESC;
```
